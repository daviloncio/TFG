## MUESTRA 50 MIL TARJETAS — 3 AL 7 FEBRERO 2025 ##

```python
import pandas as pd
pd.set_option('display.max_columns', None)

column_names = [
    'id_tarjeta',
    'cod_em_operadora',
    'fecha_validacion',
    'cod_grupo_validacion',
    'id_resultado',
    'id_tipo_titulo',
    'cod_perfil',
    'cod_descuento',
    'punto_acceso'
]

df = pd.read_csv(
    './muestra_50mil_tarjetas_3_al_7_febrero.csv',
    sep='#',
    header=None,
    names=column_names,
    parse_dates=['fecha_validacion'],
    dayfirst=True
)

df.head(10)
```

## DIMENSIONES Y TIPOS ##

```python
print(f'Filas:    {len(df):,}')
print(f'Columnas: {len(df.columns)}')
print()
df.dtypes
```

## DISTRIBUCIÓN POR OPERADORA ##

```python
df['cod_em_operadora'].value_counts().rename_axis('cod_em_operadora').reset_index(name='count')
```

## DISTRIBUCIÓN POR DÍA ##

```python
df['fecha_validacion'].dt.date.value_counts().sort_index().rename_axis('fecha').reset_index(name='count')
```

## DISTRIBUCIÓN POR GRUPO DE VALIDACIÓN ##

```python
df['cod_grupo_validacion'].value_counts().rename_axis('cod_grupo_validacion').reset_index(name='count')
```

## DISTRIBUCIÓN POR COD_PERFIL ##

```python
df['cod_perfil'].value_counts().rename_axis('cod_perfil').reset_index(name='count')
```

## TARJETAS ÚNICAS ##

```python
n_tarjetas = df['id_tarjeta'].nunique()
print(f'Tarjetas únicas:                    {n_tarjetas:,}')
print(f'Validaciones por tarjeta (media):   {len(df)/n_tarjetas:.1f}')
```
